```python
import re

def clean_text(text):
    text = str(text).lower()
    text = re.sub(r"\s+", " ", text)
    return text.strip()

df["disease_clean"] = df["disease"].apply(clean_text)
df["skin_condition_clean"] = df["skin_condition"].apply(clean_text)
df["skin_disease_clean"] = df["skin_disease"].apply(clean_text)

print("Disease fields cleaned!")
```

```python
print(df["disease_clean"].value_counts().head(30))
```

```python
df = df[
    (df["medicine_name"].str.strip() != "") &
    (
        (df["disease_clean"] != "") |
        (df["skin_condition_clean"] != "") |
        (df["skin_disease_clean"] != "")
    )
].reset_index(drop=True)

print("Usable records:", len(df))
```

```python
from sentence_transformers import SentenceTransformer

embedding_model = SentenceTransformer("all-MiniLM-L6-v2")

print("Embedding model loaded successfully!")
```

```python
embeddings = embedding_model.encode(
    df["medicine_text"].tolist(),
    batch_size=32,
    show_progress_bar=True,
    normalize_embeddings=True
)

print("Embedding shape:", embeddings.shape)
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
def recommend_medicines(disease, top_k=5):

    disease = clean_text(disease)

    query_embedding = embedding_model.encode(
        [disease],
        normalize_embeddings=True
    )

    similarity_scores = cosine_similarity(
        query_embedding,
        embeddings
    )[0]

    top_indices = similarity_scores.argsort()[-top_k:][::-1]

    results = df.iloc[top_indices].copy()

    results["similarity"] = similarity_scores[top_indices]

    return results
```

```python
def get_medicine_recommendation(disease, top_k=5):

    results = recommend_medicines(disease, top_k)

    medicines = []

    for _, row in results.iterrows():

        medicine = {
            "medicine_name": row["medicine_name"],
            "dosage_form": row["dosage_form"],
            "form": row["form"],
            "route": row["route"],
            "key_ingredients": [
                row["key_ingredients[0]"],
                row["key_ingredients[1]"],
                row["key_ingredients[2]"],
                row["key_ingredients[3]"],
                row["key_ingredients[4]"]
            ],
            "traditional_use": row["traditional_use"],
            "traditional_indication": row["traditional_indication"],
            "ayurvedic_use": row["ayurvedic_use"],
            "precautions": row["precautions"],
            "caution": row["caution"],
            "similarity": round(float(row["similarity"]), 4)
        }

        medicines.append(medicine)

    return {
        "disease": disease,
        "medicines": medicines
    }
```

```python
test_diseases = [
    "acne",
    "eczema",
    "psoriasis",
    "melasma",
    "skin pigmentation",
    "dry skin",
    "fungal skin infection",
    "itchy skin"
]

for disease in test_diseases:

    print("\n" + "=" * 60)
    print("DISEASE:", disease)
    print("=" * 60)

    results = recommend_medicines(disease, 3)

    for _, row in results.iterrows():

        print(
            row["medicine_name"],
            " | Similarity:",
            round(float(row["similarity"]), 3)
        )
```

```python
def retrieval_accuracy(k=5):

    correct = 0
    total = 0

    diseases = df["disease_clean"].dropna().unique()

    for disease in diseases:

        if disease == "":
            continue

        results = recommend_medicines(disease, k)

        returned_diseases = [
            clean_text(x)
            for x in results["disease"].tolist()
        ]

        if disease in returned_diseases:
            correct += 1

        total += 1

    if total == 0:
        return 0

    return correct / total
```

```python
accuracy = retrieval_accuracy(5)

print(
    "Top-5 Retrieval Accuracy:",
    round(accuracy * 100, 2),
    "%"
)
```
